```python
# Let's check for missing values in the dataset
```

```python
# Let's remove rows with missings value
```

```python
df = df.dropna()
```

```python
# Let's check the duplicates row in the dataset
```

```python
df.duplicated().sum()
```

```python
df['symbol'].value_counts().reset_index()
```

```python
# Let's Standardise the date format into datatime format
```

```python
df.loc[:, 'date'] = pd.to_datetime(df['date'])
```

```python
# Let us sort the dataset 
```

```python
df = df.sort_values(by=['symbol','date'])
```

```python
# Let' reset the index
```

```python
df.reset_index(drop=True, inplace=True)
```
